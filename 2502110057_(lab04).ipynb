```python
import random

accept = 0
rework = 0
scrap = 0

for i in range(100):
    diameter = random.uniform(24.90, 25.10)

    if 24.95 <= diameter <= 25.05:
        bin_name = "ACCEPT"
        accept += 1
    elif 24.90 <= diameter < 24.95 or 25.05 < diameter <= 25.10:
        bin_name = "REWORK"
        rework += 1
    else:
        bin_name = "SCRAP"
        scrap += 1

    print(f"{diameter:.2f} mm -> {bin_name}")

print("\n--- SHIFT REPORT ---")
print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)
```

IT08

```python
import random

ACCEPT_MIN, ACCEPT_MAX = 24.95, 25.05

mode = input("Select mode ('halt' to stop on scrap, 'skip' to log & continue): ").strip().lower()

accept_count = 0
rework_count = 0
scrap_count = 0
inspected = 0

batch_size = 50

print(f"\n--- Running in [{mode.upper()}] Mode ---")

for i in range(1, batch_size + 1):
    diameter = round(random.uniform(24.90, 25.10), 2)

    if ACCEPT_MIN <= diameter <= ACCEPT_MAX:
        status = "ACCEPT"
        accept_count += 1
    elif 24.92 <= diameter <= 25.08:
        status = "REWORK"
        rework_count += 1
    else:
        status = "SCRAP"
        scrap_count += 1

    inspected += 1

    if status == "SCRAP":
        if mode == "halt":
            print(f"Part {i:03d} | Diameter: {diameter:.2f} mm | Status: SCRAP --> HALTING BATCH!")
            break  # Stops the loop immediately
        elif mode == "skip":
            print(f"Part {i:03d} | Diameter: {diameter:.2f} mm | Status: SCRAP --> Logging & Skipping...")
            continue  # Skips further processing for this part and continues loop

    print(f"Part {i:03d} | Diameter: {diameter:.2f} mm | Status: {status}")

print("\n=== BATCH SUMMARY ===")
print(f"Total Inspected before termination: {inspected}")
print(f"Accepted: {accept_count} | Rework: {rework_count} | Scrap: {scrap_count}")
```

```python
import random

consecutive_accepts = 0
attempts = 0
max_attempts = 500

while consecutive_accepts < 5 and attempts < max_attempts:
    diameter = random.uniform(24.90, 25.10)
    attempts += 1

    if 24.95 <= diameter <= 25.05:
        bin_name = "ACCEPT"
        consecutive_accepts += 1
    elif 24.90 <= diameter <= 25.10:
        bin_name = "REWORK"
        consecutive_accepts = 0
    else:
        bin_name = "SCRAP"
        consecutive_accepts = 0

    print(f"Part {attempts}: {diameter:.2f} mm -> {bin_name}")

print("\n--- SHIFT REPORT ---")

if consecutive_accepts == 5:
    print("Five consecutive ACCEPTs achieved.")
    print("Parts required:", attempts)
else:
    print("Safety cap reached.")
    print("Five consecutive ACCEPTs were not achieved.")
    print("Attempts made:", attempts)
```
