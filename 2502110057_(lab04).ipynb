IT10

```python
import random

accept = 0
rework = 0
scrap = 0

for i in range(100):
    diameter = random.uniform(24.90, 25.10)

    if 24.95 <= diameter <= 25.05:
        bin_name = "ACCEPT"
        accept += 1
    elif 24.90 <= diameter < 24.95 or 25.05 < diameter <= 25.10:
        bin_name = "REWORK"
        rework += 1
    else:
        bin_name = "SCRAP"
        scrap += 1

    print(f"{diameter:.2f} mm -> {bin_name}")

print("\n--- SHIFT REPORT ---")
print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)
```

IT08

```python
import random

mode = input("Enter mode (halt/skip): ").lower()

accept = 0
rework = 0
scrap = 0
inspected = 0

for i in range(100):
    diameter = random.uniform(24.90, 25.10)
    inspected += 1

    if 24.95 <= diameter <= 25.05:
        bin_name = "ACCEPT"
        accept += 1

    elif 24.90 <= diameter <= 25.10:
        bin_name = "REWORK"
        rework += 1

    else:
        bin_name = "SCRAP"
        scrap += 1

    print(f"{diameter:.2f} mm -> {bin_name}")

    if bin_name == "SCRAP":
        if mode == "halt":
            print("SCRAP detected. Batch stopped.")
            break
        elif mode == "skip":
            print("SCRAP logged. Continuing batch...")
            continue

print("\n--- SHIFT REPORT ---")
print("Mode:", mode)
print("Inspected:", inspected)
print("ACCEPT:", accept)
print("REWORK:", rework)
print("SCRAP:", scrap)
```

```python
import random

consecutive_accepts = 0
attempts = 0
max_attempts = 500

while consecutive_accepts < 5 and attempts < max_attempts:
    diameter = random.uniform(24.90, 25.10)
    attempts += 1

    if 24.95 <= diameter <= 25.05:
        bin_name = "ACCEPT"
        consecutive_accepts += 1
    elif 24.90 <= diameter <= 25.10:
        bin_name = "REWORK"
        consecutive_accepts = 0
    else:
        bin_name = "SCRAP"
        consecutive_accepts = 0

    print(f"Part {attempts}: {diameter:.2f} mm -> {bin_name}")

print("\n--- SHIFT REPORT ---")

if consecutive_accepts == 5:
    print("Five consecutive ACCEPTs achieved.")
    print("Parts required:", attempts)
else:
    print("Safety cap reached.")
    print("Five consecutive ACCEPTs were not achieved.")
    print("Attempts made:", attempts)
```
